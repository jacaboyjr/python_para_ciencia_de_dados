## Operadores relacionais (sempre retornam um True ou False)

`>` maior que

`<` menor que

`>=` maior que ou igual

`<=` menor que ou igual

`==` comparação

`!=` diferente de



```python
# Matemática

a = 10
b = 3
c = 2

soma = a + b + c
subt = a - b - c
mult = a * b * c
div1 = a / b
div2 = a / c
potencia = a ** c
modulo_1 = a % b
modulo_2 = a % c
```

```python
print(soma, subt, mult, div1, div2, potencia, modulo_1, modulo_2)
```

```python
print(a > b)
print(a < b)
print(a <= b)
print(a >= b)
print(a != b)
print(a == b)
```

## CONDICIONAIS E OPERADORES LOGICOS

- Sempre vai ter um retorno booleano
- if, else, elif = se e se não

```python
meta = 350
venda = 300

if meta <= venda:
    print("Meta Batida")
else:
    print("Ainda não batemos a meta.")
```

```python
venda = 180

if venda >= 300:
    print("Meta Batida")
elif venda >= 200:
    print("Quase batemos a meta faltou pouco.")
elif venda >= 100:
    if venda >= 190:
        print("Voce quase chegou nos 200, vamos tentar mais um poquinho")
    else:
        print("Faltou muito para bater a meta")
else:
    print("Ainda não batemos a meta.")
```

Operadore Lógicos.

`and` = as duas condições precisam ser verdadeiras

`or` = uma das duas condições precisam ser verdadeiras

`not` = Negação se a condição for verdadeira fica falsa se for falsa fica verdadeira



```python
venda = 0

if venda > 500 or venda < 50:
    print("Hoje é um dia atípico de vendas.")
else:
    print("hoje é um dia normal de vendas.")
```

```python
venda = 600
flag = "feriado"

if flag == "feriado":
    if venda > 500:
        print("Hoje é um dia de feriado, faz sentido a nossa venda ser maior.")
    elif venda < 50:
        print("Apesar de ser feriado vendemos pouco.")
    else:
        print("Hoje foi um dia normal de vendas no feriado.")
else:
    if flag == 'normal':
        if venda > 500 or venda < 50:
            print("Vendas anormais o que pode ter acontecido?")
        else:
            print("Hoje é um dia normal de vendas.")
```

## Listas e Tuplas

- Listas são mutaveis
- Tuplas não são mutáveis

lista = `[]`
tupla = `()`

```python
frutas = ["banana", "laranja", "abacaxi", "melancia"]
frutas_vermelhas = ("amora", "franboesa", "morango", "cereja")
```

```python
print(frutas)
print(frutas_vermelhas)
```

```python
frutas.append("kiwi") #adiciona no final da lista
frutas.extend(frutas_vermelhas) # Adiciona uma lista a outra

print(frutas)
```

```python
frutas.remove("banana") # Remove um elemento especifico
frutas.pop() #remove o ultimo elemento da lista

print(frutas)
```

## Indices e Fatias

- Tudo começa em 0 da esquerda para a direira
- Tudo começa em -1 da direita para a esquerda
- Fatiamento podemos fatiar os dados pegando somente parte dele [2:5] aqui estou pegando do indice 2 até o 4 uma vez que o ultimo valor é exclusivo



```python
nomes = ["Junior", "Larissa", "Kaua", "Isadora", "Joao", "Luiza"]

lista_lista = [[1,2,3],[4,5,6],[7,8,9]]

print(nomes[1:])
print(nomes[2:-1])
print(nomes[-1::-1])

print(nomes.pop(), "- Removido !")
print(nomes)

print(lista_lista[1][2])
```
